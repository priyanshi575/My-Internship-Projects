```python
import pandas as pd

movies = {
    'title': [
        'Avatar',
        'Titanic',
        'The Notebook',
        'Inception',
        'Interstellar',
        'The Avengers',
        'Iron Man',
        'Jurassic Park',
        'Finding Nemo',
        'Frozen',
        'The Lion King',
        'Harry Potter',
        'The Conjuring',
        'Annabelle',
        'The Fault in Our Stars'
    ],

    'genre': [
        'Action Adventure Sci-Fi',
        'Romance Drama',
        'Romance Drama',
        'Action Sci-Fi Thriller',
        'Sci-Fi Drama Adventure',
        'Action Adventure Sci-Fi',
        'Action Sci-Fi',
        'Adventure Sci-Fi',
        'Animation Adventure Family',
        'Animation Family Fantasy',
        'Animation Adventure Drama',
        'Fantasy Adventure',
        'Horror Thriller',
        'Horror Thriller',
        'Romance Drama'
    ]
}

df = pd.DataFrame(movies)

df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

cv = CountVectorizer()

genre_matrix = cv.fit_transform(df['genre'])

similarity = cosine_similarity(genre_matrix)

print("Movie similarity successfully calculated!")
```

```python
def recommend_movies(movie_name):

    movie_name = movie_name.lower()

    if movie_name not in df['title'].str.lower().values:
        print("Sorry! Movie not found in our database.")
        return

    movie_index = df[df['title'].str.lower() == movie_name].index[0]

    scores = list(enumerate(similarity[movie_index]))

    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    scores = scores[1:6]

    print("\nMovies recommended for you:\n")

    for i, score in scores:
        print(df.iloc[i]['title'])

print("Recommendation function created successfully!")
```

```python
recommend_movies("Avatar")
```

```python
movie_name = input("Enter a movie name: ")

recommend_movies(movie_name)
```

**Project Workflow**

1. A movie dataset is created using Python and Pandas.
2. Movie genres are converted into numerical features using CountVectorizer.
3. Cosine Similarity is used to compare movies.
4. The user enters a movie name.
5. The system recommends five similar movies based on genre similarity.
