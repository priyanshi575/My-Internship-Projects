```python
from flask import Flask, render_template_string, request, jsonify
from threading import Thread
from google.colab import output
import time

app = Flask(__name__)

html = """
<!DOCTYPE html>
<html>
<head>
    <title>Customer Support Chatbot</title>

    <meta name="viewport" content="width=device-width, initial-scale=1.0">

    <style>
        * {
            box-sizing: border-box;
        }

        body {
            font-family: Arial, sans-serif;
            background: #f1edff;
            margin: 0;
            padding: 20px;
        }

        .container {
            max-width: 650px;
            margin: 30px auto;
            background: white;
            border-radius: 18px;
            padding: 25px;
            box-shadow: 0 5px 20px #d5cbed;
        }

        h1 {
            text-align: center;
            color: #6546c7;
        }

        .subtitle {
            text-align: center;
            color: #777;
        }

        .chatbox {
            height: 350px;
            overflow-y: auto;
            background: #f8f7ff;
            padding: 15px;
            border-radius: 12px;
            margin: 20px 0;
        }

        .message {
            padding: 12px;
            margin: 10px 0;
            border-radius: 12px;
            max-width: 85%;
            line-height: 1.5;
        }

        .bot {
            background: #e5ddff;
            color: #333;
        }

        .user {
            background: #6546c7;
            color: white;
            margin-left: auto;
        }

        .input-area {
            display: flex;
            gap: 8px;
        }

        input {
            flex: 1;
            min-width: 0;
            padding: 13px;
            border: 1px solid #ccc;
            border-radius: 8px;
            font-size: 15px;
        }

        button {
            background: #6546c7;
            color: white;
            border: none;
            padding: 12px 18px;
            border-radius: 8px;
            cursor: pointer;
        }

        button:hover {
            background: #49319a;
        }

        .suggestions {
            margin-top: 15px;
            text-align: center;
        }

        .suggestions button {
            margin: 5px;
            background: #eee9ff;
            color: #6546c7;
        }

        .footer {
            text-align: center;
            color: #777;
            margin-top: 20px;
            font-size: 13px;
        }

        @media (max-width: 480px) {
            .container {
                padding: 15px;
            }

            .chatbox {
                height: 320px;
            }
        }
    </style>
</head>

<body>

<div class="container">

    <h1>💬 Customer Support</h1>

    <p class="subtitle">
        Welcome! How can I help you today?
    </p>

    <div class="chatbox" id="chatbox">
        <div class="message bot">
            Hello! 👋 Welcome to our customer support.
            How can I assist you today?
        </div>
    </div>

    <div class="input-area">
        <input
            type="text"
            id="message"
            placeholder="Type your question..."
            onkeydown="if(event.key==='Enter') sendMessage()"
        >

        <button onclick="sendMessage()">Send</button>
    </div>

    <div class="suggestions">
        <p>Quick Questions:</p>

        <button onclick="quickAsk('Order status')">Order Status</button>
        <button onclick="quickAsk('Delivery')">Delivery</button>
        <button onclick="quickAsk('Refund')">Refund</button>
        <button onclick="quickAsk('Return')">Return</button>
        <button onclick="quickAsk('Payment')">Payment</button>
    </div>

    <div class="footer">
        Customer Support Chatbot | Available 24/7
    </div>

</div>

<script>

async function sendMessage(text) {

    let input = document.getElementById("message");
    let message = text || input.value.trim();

    if (!message) return;

    addMessage(message, "user");
    input.value = "";

    try {

        let response = await fetch("/chat", {
            method: "POST",
            headers: {
                "Content-Type": "application/json"
            },
            body: JSON.stringify({message: message})
        });

        if (!response.ok) {
            throw new Error("Server error");
        }

        let data = await response.json();
        addMessage(data.reply, "bot");

    } catch (error) {

        addMessage(
            "Sorry! Something went wrong. Please try again.",
            "bot"
        );

    }
}

function quickAsk(text) {
    sendMessage(text);
}

function addMessage(text, type) {

    let chatbox = document.getElementById("chatbox");
    let message = document.createElement("div");

    message.className = "message " + type;
    message.textContent = text;

    chatbox.appendChild(message);
    chatbox.scrollTop = chatbox.scrollHeight;

}

</script>

</body>
</html>
"""


@app.route("/")
def home():
    return render_template_string(html)


@app.route("/chat", methods=["POST"])
def chat():

    data = request.get_json(silent=True) or {}
    message = data.get("message", "").lower().strip()

    if not message:
        reply = "Please enter a message."

    elif any(word in message for word in ["hello", "hi", "hey"]):
        reply = "Hello! Welcome to our customer support. How can I help you?"

    elif "order" in message or "status" in message:
        reply = "You can check your order status using your order ID. Please contact our support team for assistance."

    elif "delivery" in message or "shipping" in message:
        reply = "Our standard delivery usually takes 3-5 business days."

    elif "refund" in message or "money back" in message:
        reply = "Refunds are generally processed within 5-7 business days after approval."

    elif "return" in message or "exchange" in message:
        reply = "You can request a return within 7 days of receiving your order."

    elif "payment" in message:
        reply = "We accept UPI, debit cards, credit cards and other online payment methods."

    elif "cancel" in message:
        reply = "For order cancellation, please contact our customer support team as soon as possible."

    elif "contact" in message or "support" in message:
        reply = "You can contact our support team at support@example.com."

    elif "thank" in message:
        reply = "You're welcome! Happy to help you."

    elif "help" in message:
        reply = "I can help with orders, delivery, refunds, returns, payments and cancellations."

    else:
        reply = "Sorry, I could not understand your question. Please ask about orders, delivery, refunds or returns."

    return jsonify({"reply": reply})


print("Customer Service Chatbot application is ready!")
```

```python
from threading import Thread
import time
from google.colab import output

def run_app():
    app.run(host="0.0.0.0", port=5002, debug=False, use_reloader=False)

server_thread = Thread(target=run_app, daemon=True)
server_thread.start()

time.sleep(3)

link = output.eval_js("google.colab.kernel.proxyPort(5002)")
print("Your Chatbot Website:", link)
```

# **Project Workflow**

1. User enters a question through the chatbot interface.
2. The message is sent to the Flask backend.
3. Python checks the message using predefined keywords.
4. The chatbot generates a relevant response.
5. The response is displayed on the website.


**Project Limitations**

This is a rule-based chatbot. It provides predefined responses and does not connect to a real order management or payment system.
